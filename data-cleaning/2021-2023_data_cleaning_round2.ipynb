```python
import pandas as pd
```

```python
file_path = "/Users/kaylamullen/Desktop/PITNE/Data/2021-2023_GEO_data.csv"
data = pd.read_csv(file_path)

# make backup copy
data_raw = data.copy()
```

```python
# Standardize column names (remove whitespace, lowercase)
data.columns = data.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
# convert objects to ints where necessary
data['hh_income_unfilled'] = (
    data['hh_income_unfilled']
    .str.replace(',', '', regex=False)
    .str.replace('$', '', regex=False)
    .str.strip()  # Remove leading/trailing spaces
)

data['hh_income_filled'] = (
    data['hh_income_filled']
    .str.replace(',', '', regex=False)
    .str.replace('$', '', regex=False)
    .str.strip()  # Remove leading/trailing spaces
)

data['hh_income_unfilled'] = pd.to_numeric(data['hh_income_unfilled'], errors='coerce')
data['hh_income_filled'] = pd.to_numeric(data['hh_income_filled'], errors='coerce')


data['hh_assets_unfilled'] = (
    data['hh_assets_unfilled']
    .str.replace(',', '', regex=False)
    .str.replace('$', '', regex=False)
    .str.strip()  # Remove leading/trailing spaces
)

data['hh_assets_filled'] = (
    data['hh_assets_filled']
    .str.replace(',', '', regex=False)
    .str.replace('$', '', regex=False)
    .str.strip()  # Remove leading/trailing spaces
)

data['hh_assets_unfilled'] = pd.to_numeric(data['hh_assets_unfilled'], errors='coerce')
data['hh_assets_filled'] = pd.to_numeric(data['hh_assets_filled'], errors='coerce')


# Convert age to numeric
data['age_unfilled'] = pd.to_numeric(data['age_unfilled'], errors='coerce')
data['age_filled'] = pd.to_numeric(data['age_filled'], errors='coerce')
```

```python
# creating asset brackets

# Define the max income to create the bins
max_assets = data['hh_assets_filled'].max()

# Create bins: from 0 to the next 20k above max_income, in steps of 20k
bins = list(range(0, int(max_assets + 20000), 20000))

# Create labels for each bin
labels = [f"${bins[i]}–${bins[i+1]-1}" for i in range(len(bins)-1)]

# Create the bracket column
data['asset_bracket'] = pd.cut(data['hh_assets_filled'], bins=bins, labels=labels, right=False)
```

```python
data['current_residence_state'] = data['current_residence_state'].str.strip()
```

```python
# Strip whitespace and replace fancy dashes or extra spaces
data['msa'] = data['msa'].str.strip()  # Remove leading/trailing spaces
data['msa'] = data['msa'].str.replace(r'\s*–\s*', '-', regex=True)  # Convert en-dash to hyphen
data['msa'] = data['msa'].str.replace(r'\s*-\s*', '-', regex=True)  # Remove spaces around hyphens

print(sorted(data['msa'].unique()))
```

```python
# refilling age column
age_median = data['age_unfilled'].median()
data['age_filled'] = data['age_unfilled'].fillna(age_median)
data['age_filled'].value_counts()
```
