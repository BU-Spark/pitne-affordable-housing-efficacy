# merging

```python
# Step 1: Add documentation type
df1['documentation'] = 'FULL'
df2['documentation'] = 'LITE'
df1_demo['documentation'] = 'FULL'
df2_demo['documentation'] = 'LITE'
```

```python
# Step 2: Ensure all expected columns exist in both
# Add missing columns to df1 with NaNs so both frames align
for col in ['fthb_class', 'hh_type', 'submission_time']:
    if col not in df1.columns:
        df1[col] = pd.NA

for col in ['fthb_class', 'hh_type', 'submission_time']:
    if col not in df1_demo.columns:
        df1_demo[col] = pd.NA
```

```python

# Optional: Reorder columns to match (makes things cleaner)
df1 = df1[df2.columns]  # assumes df2 has all the desired columns
# df1_demo = df1_demo[df2_demo.columns]  # assumes df2 has all the desired columns
```

```python
# Check if all column names match
df1_cols = set(df1.columns)
df2_cols = set(df2.columns)

print("Columns in df1 but not in df2:", df1_cols - df2_cols)
print("Columns in df2 but not in df1:", df2_cols - df1_cols)
```

```python
# Convert 0 → 'no' and 1 → 'yes'
df1['disability'] = df1['disability'].map({0: 'no', 1: 'yes'}).fillna('Choose Not To Answer')
```

```python
df1['dependents'] = df1['dependents'].astype('Int64')
df2['dependents'] = df2['dependents'].astype('Int64')
```

```python
# Find common columns
common_cols = df1_cols & df2_cols

# Compare dtypes side by side
dtype_comparison = pd.DataFrame({
    'df1_dtype': df1[list(common_cols)].dtypes,
    'df2_dtype': df2[list(common_cols)].dtypes
})

# Highlight mismatches
mismatched_dtypes = dtype_comparison[dtype_comparison['df1_dtype'] != dtype_comparison['df2_dtype']]
print("Mismatched dtypes:\n", mismatched_dtypes)
```

```python
# Step 4: Concatenate the data
merged_df = pd.concat([df1, df2], ignore_index=True)
merged_df_demo = pd.concat([df1_demo, df2_demo], ignore_index=True)
```
