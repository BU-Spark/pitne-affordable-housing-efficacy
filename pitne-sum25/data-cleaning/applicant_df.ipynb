# Creating an applican ID dataframe
Each row is a unique applicant containing their demographic info

```python
import pandas as pd
```

```python
file_path = "/Users/kaylamullen/Desktop/PITNE/Data/cleaned_data_post_GEO.csv"
df = pd.read_csv(file_path)

# make backup copy
df_raw = df.copy()
```

```python
df.head()
```

```python
duplicate_count = df['id_number'].duplicated().sum()
print(f"Number of duplicate applicant IDs: {duplicate_count}")
```

```python
# 1. Make sure submission_date is in datetime format
df['submission_date'] = pd.to_datetime(df['submission_date'])
```

```python
df.head()
```

```python
# 2. Sort by applicant_id and submission_date
df_sorted = df.sort_values(by=['id_number', 'submission_date'])
```

```python
# 3. Drop duplicates, keeping the earliest application per applicant
unique_applicants = df_sorted.drop_duplicates(subset='id_number', keep='first')
```

```python
print(unique_applicants.shape)
unique_applicants.head()
```

```python
duplicate_count = unique_applicants['id_number'].duplicated().sum()
print(f"Number of duplicate applicant IDs: {duplicate_count}")
```

# Save to csv

```python
unique_applicants.to_csv('/Users/kaylamullen/Desktop/PITNE/Data/2021-2023_unique_applicants.csv', index=False)
```
