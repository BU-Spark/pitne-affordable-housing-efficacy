# Dependents

## Dependents and Bedrooms

```python
#Average Bedrooms per Town
#21-23
# Make sure 'bedrooms' is numeric
df1['bedrooms'] = pd.to_numeric(df1['bedrooms'], errors='coerce')

# Group by town and compute average number of bedrooms
avg_bedrooms_by_town = (
    df1.groupby('property_town')['bedrooms']
    .mean()
    .dropna()
    .sort_values(ascending=False)
)

print(avg_bedrooms_by_town)
```

```python
#23-25

df2['bedrooms'] = pd.to_numeric(df2['bedrooms'], errors='coerce')

# Group by town and compute average number of bedrooms
avg_bedrooms_by_town = (
    df2.groupby('property_town')['bedrooms']
    .mean()
    .dropna()
    .sort_values(ascending=False)
)
print(avg_bedrooms_by_town)
```

```python
# Combine the two DataFrames
df = pd.concat([df1, df2], ignore_index=True)

mpl.rcParams['font.family'] = 'Lexend'

# Create crosstab of bedroom vs dependents
bedroom_dep_ct = pd.crosstab(df['bedrooms'], df['dependents'])

# Plot heatmap
plt.figure(figsize=(10, 6))
ax = sns.heatmap(bedroom_dep_ct, annot=True, fmt='d', cmap= custom_gradient)
plt.title('Number of Applicants by Bedrooms vs. Dependents - Combined Data')
plt.xlabel('Number of Dependents')
plt.ylabel('Bedrooms Applied For')
ax.invert_yaxis()
plt.tight_layout()
plt.savefig('/Users/saniyajain/Downloads/CHAPA Saniya DS/Figures /BQ1/num_bedrooms_dependents_merged.png', dpi=300)
plt.show()
```

## Dependents Across Towns

### 23-25 

```python
custom_gradient = LinearSegmentedColormap.from_list(
    name='custom_gradient',
    colors=[ '#005757', '#daf3df'], 
    N=10  
)

# Filter for applicants with one or more dependents
df_dependents = df2[df2['dependents'].fillna(0).astype(int) > 0]

# Count applications by town
town_counts = df_dependents['property_town'].value_counts().head(10)

# Convert counts to percentage of total dependent applications
total_dependents = df_dependents.shape[0]
town_percentages = (town_counts / total_dependents) * 100

# Plot
plt.figure(figsize=(12, 6))
sns.barplot(
    x=town_percentages.index,
    y=town_percentages.values,
    palette=[custom_gradient(i / 9) for i in range(10)]
)

plt.title('2023-25: Top 10 Towns by Percentage of Dependent Applicants', fontsize=14)
plt.xlabel('Property City/Town', fontsize=12)
plt.ylabel('Percentage of Applicants with Dependents (%)', fontsize=12)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.tight_layout()
plt.savefig('/Users/saniyajain/Downloads/CHAPA Saniya DS/Figures /BQ1/top_10_towns_dependents_percentages23.png', dpi=300)
plt.show()
```

# FTHB

## Proportion of FTBH and Non By Town

```python

df2['fthb_class'] = df2['fthb_class'].str.strip().str.lower()


fthb_counts = df2.groupby(['property_town', 'fthb_class']).size().unstack(fill_value=0)

# Rename columns for clarity
fthb_counts.columns = ['Non-FTHB' if col != 'yes' else 'FTHB' for col in fthb_counts.columns]

# Get top 10 towns by total applications
top_10_towns = fthb_counts.sum(axis=1).sort_values(ascending=False).head(10).index
fthb_top10 = fthb_counts.loc[top_10_towns]

fthb_top10_percentage = fthb_top10.div(fthb_top10.sum(axis=0), axis=1) * 100

#  Plot
plt.figure(figsize=(14, 7), dpi = 250)
fthb_top10_percentage.plot(
    kind='bar',
    stacked=False,
    width=0.8,
    color=['#ff8f80', '#005757']
)

plt.title('Top 10 Towns: FTHB vs Non-FTHB Applicants', fontsize=14)
plt.xlabel('Town', fontsize=12)
plt.ylabel('Proportion of Applicants', fontsize=12)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.legend(title='Applicant Type')
plt.tight_layout()
plt.savefig('/Users/saniyajain/Downloads/CHAPA Saniya DS/Figures /BQ1/top_10_fthb_vs_nonfthb_percentages.png', dpi=300)
plt.show()
```

```python
fthb_counts = df2.groupby(['property_town', 'fthb_class']).size().unstack(fill_value=0)
fthb_counts.columns = ['Non-FTHB' if col != 'yes' else 'FTHB' for col in fthb_counts.columns]

# Get top 10 towns by total application count
top_10_towns = fthb_counts.sum(axis=1).sort_values(ascending=False).head(10).index
fthb_top10 = fthb_counts.loc[top_10_towns]

# Calculate within-town percentage of FTHBs
fthb_top10_percent_within = fthb_top10['FTHB'] / fthb_top10.sum(axis=1) * 100

# Plot
plt.figure(figsize=(12, 6), dpi=250)
sns.barplot(x=fthb_top10_percent_within.index, y=fthb_top10_percent_within.values, color='#005757')


plt.title('Percentage of FTHB Applicants Within Top 10 Towns', fontsize=14)
plt.xlabel('Town', fontsize=12)
plt.ylabel('Percent of Applicants Who Are FTHBs (%)', fontsize=12)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.ylim(0, 100)
plt.tight_layout()
plt.savefig('/Users/saniyajain/Downloads/CHAPA Saniya DS/Figures /BQ1/percent_fthb_top_ten.png', dpi=250)
plt.show()
```

## % of FTHB by Town (what percentage of all apps were FTHB)

```python

# Count total applicants per town
total_by_town = df2.groupby('property_town').size()

# Count number of first-time home buyers ("Yes") per town
fthb_by_town = df2[df2['fthb_class'].str.lower() == 'yes'].groupby('property_town').size()

#Calculate percentage
fthb_percent = (fthb_by_town / total_by_town) * 100

# Format into a DataFrame
fthb_percent_df = fthb_percent.reset_index()
fthb_percent_df.columns = ['City/Town', 'First Time Home Buyer %']

fthb_percent_df = fthb_percent_df.sort_values(by='First Time Home Buyer %', ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(
    data=fthb_percent_df,
    x='City/Town',
    y='First Time Home Buyer %',
    color='#005757'
)
plt.title('Percentage of First-Time Home Buyers by Town')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('/Users/saniyajain/Downloads/CHAPA Saniya DS/Figures /BQ1/percentage_of_FTHB_by_town.png', dpi=250)
plt.show()
```
