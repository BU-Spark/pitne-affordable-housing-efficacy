Write a script that:
1. gets the number of rows with empty closing dates

2. finds rows with Transaction Start Date < 09/01/2024 (September 1st 2024) 
and closing date is empty/null 

3. Gets the number of such rows

4. Marks their Closing Date as 1900-09-01 for now

(This is flagged for our clients.)

```python
def update_Closing_Date_Create_Duplicate_csv(df):
    df['Transaction Start Date'] = pd.to_datetime(df['Transaction Start Date'])
    df['Closing Date'] = pd.to_datetime(df['Closing Date'])
    # changed formats
    print("The number of rows with no Closing Dates")
    df[(df['Closing Date'].isna())].count()
    print("The number of rows with no Closing Dates and Transaction Start Date Before Spetember 01 2024")
    df[(df['Closing Date'].isna()) & (df['Transaction Start Date']<'2024-09-01')].count()
    # printing values
    df.loc[(df['Closing Date'].isna()) & (df['Transaction Start Date']<'2024-09-01'), 'Closing Date'] = pd.Timestamp('1900-09-01')
    df.to_csv("duplicate_round2.csv", index=False)
    # making the change and creating a duplicate csv in the directory
    return df
```

Fill the number of rows with Age Restricted values as null to ‘No’
Change that to null according to information from client

```python
def update_age_restricted_Create_Duplicate_csv(df):
    #the ones with 'No'
    df[(df['Age Restricted'] == 'No')].describe()
    #the ones with null
    df[(df['Age Restricted'].isna())].describe()
    #standardizing
    df['Age Restricted'] = df['Age Restricted'].fillna('No')
    #creating duplicate csv
    df.to_csv("duplicate_round2.csv", index=False)
    return df
```
