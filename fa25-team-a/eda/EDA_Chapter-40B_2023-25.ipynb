## Manipulate dataset


ask client about possible `hh_income` outliers

* ask how to handle:
    * `age`: "0", "4", what is age requirement to apply?
    * `race_combination_normalized`: drop "Missing" (3), "Other" (1)?
    * use raw data from `race_combination_normalized` or still adhere to `census_races`?
        * disregards "Hispanic or Latino"
    * "Choose not to answer" in `race_combination_normalized`, `disability`
    * `hh_size`: "20", "31"
    * `hh_type_standardized`: "Other" (13)
    * does CHAPA only look at `hh_income` or `hh_income` + `hh_assets`?
    * limited nulls: `age`, `disability`, `current_residence_town_city`, `current_residence_state`, `hh_income`, `hh_assets`, `fthb_class`

* clean:
    * `application_unit`: filters out "unit" or "#"
    * `race_combination_normalized`: consider changing to list dtype
    * `disability`: standardize "no" to "No"
    * `current_residence_state`: standardize "Massachusetts", "ma", "Ma" to "MA"

* create:
    * `age_restricted`: 1 if indicated as "(55+)" or "(age restricted)" in `application_property`, 0 otherwise
    * `zip`: pull from `matched_address`



```python
# standardize columns
# disability
df["disability"] = df["disability"].replace({"no": "No"})
# current_residence_state
for x in ["Ma", "ma", "Massachusetts"]:
    df["current_residence_state"] = df["current_residence_state"].replace({x: "MA"})
# application_unit
df["application_unit"] = (
    df["application_unit"]
    .str.replace(r'(?i)unit', '', regex=True)   #case-insensitive
    .str.replace('#', '', regex=False)
    .str.strip()
)
```
