## Data Preprocessing & EDA

```python
# Missing Age was replaced by the median value of 'Age' by the Summer 2025 team.
# Decided to round: age_median = 38.5 -> 39

for i in range(len(df)):
  if df['Age Filled'][i] == '38.5':
    df.loc[i, 'Age Filled'] = 39

print(df['Age Filled'])
```

```python
# Change the data type 'object' to 'float'
# Visualize as a histogram for better comparison.

df['HH Income Filled'] = df['HH Income Filled'].str.replace(r'\.00$', '', regex=True)
df['HH Income Filled'] = df['HH Income Filled'].str.replace(',', '').astype(float)
print(df['HH Income Filled'])
```

```python
# Change the data type 'object' to 'float'
# Visualize as a histogram for better comparison.

df['HH Assets Filled'] = df['HH Assets Filled'].str.replace(r'\.00$', '', regex=True)
df['HH Assets Filled'] = df['HH Assets Filled'].str.replace(',', '').astype(float)
print(df['HH Assets Filled'])
```

```python
# Fill 'ZIP' column by importing zip code from 'Matched Address' column

for i in range(len(df)):
  a = df['Matched Address'][i].split(",")[-1]
  space_count = a.count(" ")
  if space_count > 0:
    b = a.split(" ")[-1]
    df.loc[i, 'ZIP'] = b
```

```python
# Changing the data type of 'ZIP' as string to get '0' as first number.

df['ZIP'] = df['ZIP'].astype(str)
print(df['ZIP'].head(15))
```

```python
# One unknown value 'u/a' in 'Age Unfilled' -> replaced with the age_median '39'

a = df['Age Unfilled']
a = a.drop([294]) # a[294] = 'u/a'
a = a.dropna()
age_median = a.astype(int).median()
df.loc[294, 'Age Filled'] = age_median
```

```python
df['new_range'] = df['Income Bracket'].str.replace('$', '').str.replace(',', '').str.replace('–', '-')
def convert_to_k(x):
  start, end = x.split('-')
  start_k = int(start) // 1000
  end_k = int(end) // 1000
  return f'{start_k}k-{end_k}k'

def convert(x):
  start, end = x.split('-')
  start_k = start[:-3] + 'k'
  end_k = end[:-3] + 'k'
  return f'{start_k}-{end_k}'

df['Income Bracket in K'] = df['new_range'].apply(convert)
print(df['Income Bracket in K'])
```

```python
sorting_order = ['$0k-19k', '$20k-39k', '$40k-59k', '$60k-79k', '$80k-99k', '$100k-119k']
df['Income Bracket in K'] = pd.Categorical(df['Income Bracket in K'], categories=sorting_order, ordered=True)

range_counts = df['Income Bracket in K'].value_counts().sort_index()

plt.figure(figsize=(10, 6))
range_counts.plot(kind='bar', rot=0)
plt.title('Income Range Distribution')
plt.xlabel('Income Range')
# plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Plotting object type of column as numeric values
range_counts = df['Income Bracket'].value_counts()

# Change data type
sorted_indices = range_counts.index.to_series().str.replace('$', '', regex=False).str.replace(',', '', regex=False).sort_values(
    key=lambda x: x.str.split('–').str[0].astype(int),
    ascending=True
).index

sorted_counts = range_counts.reindex(sorted_indices)

plt.figure(figsize=(10, 6))
sorted_counts.plot(kind='bar', rot=45)
plt.title('Income Bracket Distribution')
plt.xlabel('Income Bracket')
# plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Summary

1. Household and Demographic Statistics

* Age Data Quality: The percentage of missing values for Age comprises a significant portion of the total, approximately one-third of the dataset. Although the median age of 39 was used as an initial substitute for primary analysis, this requires further discussion and refinement in consultation with the client.
* Household Size and Composition: Approximately 45% of applicants are one-person households. This shows that low-occupancy households (especially 1-2 person households) are primarily seeking affordable housing options, with most having no dependents.


2. Income and Asset Statistics

* Household Income (HH Income): The distribution of Household Income is highly skewed to the right, indicating that the majority of applicants have lower incomes. This reflects a general characteristic of individuals searching for affordable housing.
* Income Bracket Concentration: Income is heavily concentrated in the 40k–59k and 60k–79k brackets. These two brackets account for over 76% of all applications, clearly showing that applicants fall into the middle-to-lower income categories in Massachusetts.
* IHousehold Assets (HH Assets): The distribution of Household Assets is also skewed to the right, suggesting that most applicants have low asset holdings, with only a few reporting high asset values. The most common asset brackets observed are 20k–40k, 40k–60k, and 60k–100k. This implies that many applicants may be asset-poor relative to their income.


3. Race/Ethnicity and Geographic Statistics

* Race/Ethnicity Demographics: The largest group in the applications is White/Non-Minority (CHAPA definition) or White alone (Census definition). The differing classification methods for the Hispanic/Latino category between the 'CHAPA Race/Ethnicity' and 'Census Race' columns requires clarification and potential reconciliation with the client.
* Geographic Distribution: All applied properties are located in MA. While most applicants reside in MA, a minority are currently residents of other states, including CT, NH, and RI. In-depth analysis of geographic data through geocoding is warranted to understand migration patterns and regional demand.


※ Summary and Future Analysis Direction

* The majority of applicants are within the economically active age range and primarily constitute small households (1–2 persons) actively seeking affordable housing. However, the high volume of missing age data necessitates follow-up action.
* The applicant pool appears to contain a significant number of individuals who are asset-poor relative to their income level.
