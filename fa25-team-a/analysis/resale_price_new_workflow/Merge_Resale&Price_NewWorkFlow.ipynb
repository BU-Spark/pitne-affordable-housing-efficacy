# Merge Updated Price Dataset with Resale Data

This notebook merges the updated price dataset (original price + completed missing prices)  
with the resale dataset (`ResaleCombined_Sept25.csv`) to create a fully integrated dataset for analysis.

This step occurs **after updating the original price dataset with completed missing prices**.

### Pipeline Overview
1. After running `ParsePrices_NewWorkFlow.ipynb` (found at: fa25-team-a/analysis/resale_price_new_workflow/ParsePrices_NewWorkFlow.ipynb) and get the updated price dataset (`PriceData_Filled.csv`)  
2. Load resale dataset (`ResaleCombined_Sept25.csv`)  
3. Merge datasets on `Town`, `Address`, and `Unit Number`  
   - Use `outer` join to retain all properties  
4. Save the fully merged dataset (`MergedPrice_Resale_Full.csv`)


```python
# ============================================================
# CHAPA Data Integration Pipeline
# Step: Merge Updated Price Data with Resale
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1. Load Datasets
# ------------------------------------------------------------
def load_datasets():
    """
    Load updated price dataset (original + missing prices filled)
    and resale dataset.
    """
    updated_price = pd.read_csv("../../data/PriceData_Filled.csv")
    resale_df = pd.read_csv("../../data/ResaleCombined_Sept25.csv")
    return updated_price, resale_df

# ------------------------------------------------------------
# 2. Merge Datasets
# ------------------------------------------------------------
def merge_price_resale(updated_price, resale_df):
    """
    Merge updated price data with resale dataset.
    Outer join ensures all rows are preserved.
    """
    keys = ["Town", "Address", "Unit Number"]
    merged = pd.merge(updated_price, resale_df, on=keys, how="outer")
    return merged

# ------------------------------------------------------------
# 3. Save Merged Dataset
# ------------------------------------------------------------
def save_merged_dataset(df, filename="MergedPrice_Resale_Full.csv"):
    """Save the fully merged price and resale dataset."""
    df.to_csv(filename, index=False)
    print(f"✅ Fully merged dataset saved as {filename}")

# ------------------------------------------------------------
# 4. Pipeline Execution
# ------------------------------------------------------------
def main():
    print("🚀 Starting merge of updated price data with resale...")
    updated_price, resale_df = load_datasets()
    merged_df = merge_price_resale(updated_price, resale_df)
    save_merged_dataset(merged_df)
    print("🏁 Merge complete. Dataset ready for analysis.")

if __name__ == "__main__":
    main()
```
