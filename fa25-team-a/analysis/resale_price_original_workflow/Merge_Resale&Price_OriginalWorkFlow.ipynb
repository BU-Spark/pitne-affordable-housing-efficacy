# Merging Original Price Data with Resale Data (Pre-Missing Prices)

This notebook merges the parsed original price dataset (`Jun21_Sept25_Parsed.csv`)  
with the resale dataset (`ResaleCombined_Sept25.csv`) to create a combined dataset for further analysis.  

This step occurs **before integrating completed missing price data**, producing the dataset  
`merged_dataset_Jun21_Sept25.csv`.

### Pipeline Overview
1. Load parsed price dataset (June 2021 – September 2025)  
2. Load resale dataset (`ResaleCombined_Sept25.csv`)  
3. Merge datasets on `Town`, `Address`, and `Unit Number`  
   - Use `outer` join to retain all properties from both datasets  
4. Save the merged dataset (`merged_dataset_Jun21_Sept25.csv`)  


```python
# ============================================================
# CHAPA Price & Resale Merge (Pre-Missing Prices)
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1. Load Datasets
# ------------------------------------------------------------
def load_datasets():
    """Load parsed original price dataset and resale dataset."""
    price_df = pd.read_csv("../../data/Jun21_Sept25_Parsed.csv")
    resale_df = pd.read_csv("../../data/ResaleCombined_Sept25.csv")
    return price_df, resale_df

# ------------------------------------------------------------
# 2. Merge Datasets
# ------------------------------------------------------------
def merge_price_resale(price_df, resale_df):
    """
    Merge parsed price data with resale data on key columns.
    Outer join ensures all rows are preserved from both datasets.
    """
    keys = ["Town", "Address", "Unit Number"]
    merged = pd.merge(price_df, resale_df, on=keys, how="outer")
    return merged

# ------------------------------------------------------------
# 3. Save Merged Dataset
# ------------------------------------------------------------
def save_merged_dataset(df, filename="merged_dataset_Jun21_Sept25.csv"):
    """Save the merged dataset to CSV."""
    df.to_csv(filename, index=False)
    print(f"✅ Merged dataset saved as {filename}")

# ------------------------------------------------------------
# 4. Pipeline Execution
# ------------------------------------------------------------
def main():
    print("🚀 Starting price & resale merge pipeline...")
    price_df, resale_df = load_datasets()
    merged_df = merge_price_resale(price_df, resale_df)
    save_merged_dataset(merged_df)
    print("🏁 Merge complete. Dataset ready for missing prices integration.")

if __name__ == "__main__":
    main()
```
