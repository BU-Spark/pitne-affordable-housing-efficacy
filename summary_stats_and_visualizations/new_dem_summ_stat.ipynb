# Race

```python
mpl.rcParams['font.family'] = 'Lexend'

race_counts = df1['race_simplified'].value_counts()
race_labels = race_counts.index
race_sizes = race_counts.values

colors = [source_color_map.get(label, '#CCCCCC') for label in race_labels]

# show percentage labels only for slices >3%
def autopct_func(pct):
    return f'{pct:.1f}%' if pct > 3 else ''

plt.figure(figsize=(8, 4), dpi=250)
plt.figure(figsize=(8, 4), dpi=250)
wedges, _ , _ = plt.pie(
    race_sizes,
    labels=None,
    autopct=autopct_func,
    startangle=140,
    counterclock=False,
    textprops={'fontsize': 12, 'color': 'white'},
    wedgeprops={'linewidth': 0},
    colors=colors
)

legend_items = sorted(zip(race_labels, wedges), key=lambda x: x[0]) 
sorted_labels, sorted_wedges = zip(*legend_items)

# Add sorted legend
plt.title("Race/Ethnicity Applicant Distribution - 21", fontsize=16)
plt.legend(sorted_wedges, sorted_labels, title="Race/Ethnicity", bbox_to_anchor=(1, 0.5), loc="center left", fontsize=11, title_fontsize=12)


plt.axis('equal')  # Keep it circular
plt.tight_layout()
plt.savefig('/Users/saniyajain/Downloads/CHAPA Saniya DS/Figures /BQ3/race_21.png', dpi=250)
plt.show()
```

# Age Restricted and Non Age Restricted Demographic Analysis

To ensure that the demographic analysis does not have duplicate applicants, creating separate dataframes with only first entries

```python
# Merge 21 and 23 datasets for age restricted and non age restricted 
# Merge age-restricted datasets
df_age_full = pd.concat([df1_age, df2_age], ignore_index=True)

# Merge non-age-restricted datasets
df_non_full = pd.concat([df1_non, df2_non], ignore_index=True)
```

```python
df_age_full['id_number'].duplicated().sum()
```

```python
# Keep first occurrence of each duplicate id_number
df_age_full_dem = df_age_full.drop_duplicates(subset='id_number', keep='first')
df_non_full_dem = df_non_full.drop_duplicates(subset='id_number', keep='first')
```

Creating datasets to use for maps 

```python
df_age_full.to_csv('../ArcGis/age_restricted_merged.csv', index = False )
df_non_full.to_csv('../ArcGis/non_age_restricted_merged.csv', index= False )
```

```python
# For Age-Restricted Applications
age_counts = df_age_full['property_town'].value_counts().reset_index()
age_counts.columns = ['property_town', 'app_count']
age_counts['percentage'] = (age_counts['app_count'] / age_counts['app_count'].sum()) * 100
age_counts.to_csv('../ArcGis/age_restricted_percentages.csv', index=False)

# For Non-Age-Restricted Applications
non_counts = df_non_full['property_town'].value_counts().reset_index()
non_counts.columns = ['property_town', 'app_count']
non_counts['percentage'] = (non_counts['app_count'] / non_counts['app_count'].sum()) * 100
non_counts.to_csv('../ArcGis/non_age_restricted_percentages.csv', index=False)
```
