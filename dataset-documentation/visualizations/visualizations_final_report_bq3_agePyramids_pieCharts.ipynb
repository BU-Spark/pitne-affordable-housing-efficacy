# function for creating pie chart

```python
def make_autopct_top_n_by_pct(values, top_n=4):
    total = sum(values)
    percentages = [(v / total) * 100 for v in values]

    # Get indices of top N percentages
    top_indices = sorted(range(len(percentages)), key=lambda i: percentages[i], reverse=True)[:top_n]

    def autopct(pct):
        nonlocal current_index
        label = f'{pct:.1f}%' if current_index in top_indices else ''
        current_index += 1
        return label
    current_index = 0
    return autopct
```

```python
import matplotlib.patches as mpatches

def createComparisonPlot(counts_1, labels_1, counts_2, labels_2, title, font_size = 10):
    labels_1 = list(map(str.title, labels_1))
    labels_2 = list(map(str.title, labels_2))
    fig = plt.figure(figsize=(10, 5), dpi=250)

    # First Pie
    ax1 = plt.subplot(1, 2, 1)
    autopct1 = make_autopct_top_n_by_pct(counts_1)
    ax1.pie(
        counts_1,
        colors=colors[:len(counts_1)],
        autopct=autopct1,
        pctdistance=0.7,
        textprops={'fontsize': font_size, 'color': 'white'},
        wedgeprops={'edgecolor': 'black', 'linewidth': 1.2}
    )
    ax1.set_title("2021–2023", fontsize=14, color=text_color)

    # Create proxy legend handles for first pie
    handles1 = [mpatches.Patch(color=colors[i]) for i in range(len(counts_1))]
    ax1.legend(handles1, labels_1, loc='center left', bbox_to_anchor=(1, 0.9), fontsize=9)

    # Second Pie
    ax2 = plt.subplot(1, 2, 2)
    autopct2 = make_autopct_top_n_by_pct(counts_2)
    ax2.pie(
        counts_2,
        colors=colors[:len(counts_2)],
        autopct=autopct2,
        pctdistance=0.7,
        textprops={'fontsize': font_size, 'color': 'white'},
        wedgeprops={'edgecolor': 'black', 'linewidth': 1.2}
    )
    ax2.set_title("2023–2025", fontsize=14, color=text_color)

    # Create proxy legend handles for second pie
    handles2 = [mpatches.Patch(color=colors[i]) for i in range(len(counts_2))]
    ax2.legend(handles2, labels_2, loc='center left', bbox_to_anchor=(1, 0.9), fontsize=9)

    # Centered Title
    plt.suptitle(title, fontsize=15, color=text_color)

    plt.tight_layout()
    plt.subplots_adjust(top=0.85)
    plt.show()
```

# Age Distribution

```python
# df_2021['age'].value_counts()
# # df_2023['age'].value_counts()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Clean and convert age columns
df_2021['age'] = pd.to_numeric(df_2021['age'], errors='coerce').dropna().astype(int)
df_2023['age'] = pd.to_numeric(df_2023['age'], errors='coerce').dropna().astype(int)

# Define 5-year bins
min_age = int(min(df_2021['age'].min(), df_2023['age'].min()))
max_age = int(max(df_2021['age'].max(), df_2023['age'].max()))
bins = list(range((min_age // 5) * 5, ((max_age // 5) + 1) * 5 + 1, 5))
labels = [f"{b}-{b+4}" for b in bins[:-1]]

# Assign age bins
df_2021['age_group'] = pd.cut(df_2021['age'], bins=bins, labels=labels, right=False)
df_2023['age_group'] = pd.cut(df_2023['age'], bins=bins, labels=labels, right=False)

# Calculate percentages
age_counts_2021 = df_2021['age_group'].value_counts(normalize=True).sort_index() * 100
age_counts_2023 = df_2023['age_group'].value_counts(normalize=True).sort_index() * 100

# Filter to include only age groups starting at 20
filtered_labels = [label for label in labels if int(label.split('-')[0]) >= 20]
left_data = -age_counts_2021[filtered_labels]
right_data = age_counts_2023[filtered_labels]

# Plot
fig, ax = plt.subplots(figsize=(10, 6), dpi=250)

# Bars
ax.barh(filtered_labels, left_data, color='#ff8f80', label='FULL')   # Coral
ax.barh(filtered_labels, right_data, color='#005757', label='LITE')  # CHAPA Green

# Add percentage labels with adjusted padding
label_padding = 0.5

for i, label in enumerate(filtered_labels):
    ax.text(left_data[i] - label_padding, i, f"{abs(left_data[i]):.1f}%",
            va='center', ha='right', fontsize=12, color='#1f1e40')
    ax.text(right_data[i] + label_padding, i, f"{right_data[i]:.1f}%",
            va='center', ha='left', fontsize=12, color='#1f1e40')

# X-axis ticks: from 1% to 25% on both sides
ticks_pos = np.arange(1, 26, 3)
xticks = np.concatenate([-ticks_pos[::-1], ticks_pos])
ax.set_xticks(xticks)
ax.set_xticklabels([f"{abs(int(x))}%" for x in xticks])
ax.set_xlim(-25, 25)

# Labels and style
ax.set_xlabel('Percentage of Applicants (%)', fontsize=14, color='#1f1e40')
ax.set_ylabel('Age Group (5-Year Bins)', fontsize=14, color='#1f1e40')
ax.set_title('Age Distribution Pyramid: FULL vs LITE', fontsize=20, color='#1f1e40', weight='bold')
ax.tick_params(axis='both', labelsize=12, colors='#1f1e40')
ax.legend(loc='upper right', fontsize=12)
plt.tight_layout()
plt.show()
```

# Age Distribution Pyramid Function Test

```python
def createAgePyramid(df1, df2, title_text, label_df1, label_df2, min_age, xlim_max, x_skip_interval = 3):
    # Convert age columns to integers
    df1['age'] = pd.to_numeric(df1['age'], errors='coerce').dropna().astype(int)
    df2['age'] = pd.to_numeric(df2['age'], errors='coerce').dropna().astype(int)

    # Determine bin range
    overall_min_age = int(min(df1['age'].min(), df2['age'].min()))
    overall_max_age = int(max(df1['age'].max(), df2['age'].max()))
    bin_start = (overall_min_age // 5) * 5
    bin_end = ((overall_max_age // 5) + 1) * 5
    bins = list(range(bin_start, bin_end + 1, 5))
    labels = [f"{b}-{b+4}" for b in bins[:-1]]

    # Assign age bins
    df1['age_group'] = pd.cut(df1['age'], bins=bins, labels=labels, right=False)
    df2['age_group'] = pd.cut(df2['age'], bins=bins, labels=labels, right=False)

    # Calculate percentages
    age_counts_1 = df1['age_group'].value_counts(normalize=True).sort_index() * 100
    age_counts_2 = df2['age_group'].value_counts(normalize=True).sort_index() * 100

    # Filter labels based on passed-in min_age
    filtered_labels = [label for label in labels if int(label.split('-')[0]) >= min_age]
    left_data = -age_counts_1[filtered_labels]
    right_data = age_counts_2[filtered_labels]

    # Create plot
    fig, ax = plt.subplots(figsize=(10, 6), dpi=250)

    # Bars
    ax.barh(filtered_labels, left_data, color='#ff8f80', label=label_df1)   # Coral
    ax.barh(filtered_labels, right_data, color='#005757', label=label_df2)  # CHAPA Green

    # Add percentage labels
    label_padding = 0.5
    for i, label in enumerate(filtered_labels):
        ax.text(left_data[i] - label_padding, i, f"{abs(left_data[i]):.1f}%",
                va='center', ha='right', fontsize=12, color='#1f1e40')
        ax.text(right_data[i] + label_padding, i, f"{right_data[i]:.1f}%",
                va='center', ha='left', fontsize=12, color='#1f1e40')

    # Determine dynamic tick step based on figure width and xlim_max
    fig_width = fig.get_size_inches()[0]
    approx_labels_allowed = fig_width * 1.5  # heuristic: ~1.5 labels per inch
    max_ticks = int(approx_labels_allowed)

    # Find appropriate tick step (at least 1)
    step = max(1, int(np.ceil(xlim_max / (max_ticks / 2))))

    # Generate symmetric tick positions
    ticks_pos = np.arange(step, xlim_max + 1, step)
    xticks = np.concatenate([-ticks_pos[::-1], ticks_pos])
    ax.set_xticks(xticks)
    ax.set_xticklabels([f"{abs(int(x))}%" for x in xticks])
    ax.set_xlim(-xlim_max, xlim_max)


    # Labels and styling
    ax.set_xlabel('Percentage of Applicants (%)', fontsize=13, color='#1f1e40')
    ax.set_ylabel('Age Group (5-Year Bins)', fontsize=14, color='#1f1e40')
    ax.set_title(title_text, fontsize=20, color='#1f1e40', weight='bold')
    ax.tick_params(axis='both', labelsize=12, colors='#1f1e40')
    ax.legend(loc='upper right', fontsize=12)

    plt.tight_layout()
    plt.show()
    return
```
