## Creating Dataset for proportion of apps with dependents to each town

```python
apps_with_dependents = df_merged_dependents.groupby('property_town')['id_number'].count().reset_index()
apps_with_dependents.columns = ['property_town', 'apps_with_dependents']
```

```python
total_dependents_apps = df_merged_dependents['id_number'].count()
```

```python
apps_with_dependents['dependents_ratio'] = apps_with_dependents['apps_with_dependents'] / total_dependents_apps
apps_with_dependents['dependents_percentage'] = apps_with_dependents['dependents_ratio'] * 100
```

```python
apps_with_dependents.to_csv('dependent_apps_percent_by_town.csv', index=False)
```

```python
apps_with_dependents.sort_values(by='dependents_percentage', ascending=False, inplace=True)
```
