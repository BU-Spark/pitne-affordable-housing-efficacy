# Load Datasets and Imports 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

dfr = pd.read_csv("Resale_transaction_data.csv")
dfr.head()
```

# Mergin 23-25 data with Resale Data 

```python
df = pd.read_csv("data_updated.csv") 
```

```python
dfr.columns
```

```python
df[['Application City/Town', 'Application St Address', 'Application Unit']].head()
```

```python
# only keep first occurence of macthed street address from dfr 
dfr_deduped = dfr.drop_duplicates(subset='Matched Address', keep='first')

df_merged = pd.merge(
    df,
    dfr_deduped[['Matched Address', 'Bedrooms', 'Age Restricted']],
    on='Matched Address',
    how='left',
    indicator=True
)

```

```python
num_matched = (df_merged['_merge'] == 'both').sum()
print(f"Number of matched rows: {num_matched}")
```

```python
df_merged.shape
```

```python
# Export to CSV
df_merged.to_csv('../Summary Stats/23_25_bedroom_age.csv', index=False)
```

# Merging 21-23 data with Resale Data

```python
dfs = pd.read_csv('clean_2021_2023_applications.csv')
```

```python
dfs.columns
```

```python
# Only keep first occurrence of Matched Address from dfr
dfr_deduped = dfr.drop_duplicates(subset='Matched Address', keep='first')

# Merge on differently named address columns
df_merged_21 = pd.merge(
    dfs,
    dfr_deduped[['Matched Address', 'Bedrooms', 'Age Restricted']],
    left_on='matched_address',
    right_on='Matched Address',
    how='left',
    indicator=True
)
```

```python
num_matched_21 = (df_merged_21['_merge'] == 'both').sum()
print(f"Number of matched rows: {num_matched_21}")
```

```python
# export to csv
df_merged_21.to_csv('../Summary Stats/21_23_bedroom_age.csv', index=False)
# Check the shape of the merged DataFrame
df_merged_21.shape
```
